```python
def union(arr1,arr2):
    n, m = len(arr1), len(arr2)
    i = 0
    j = 0
    un=[]
    def add(element):
        if not un or un[-1]!=element:
            un.append(element)
    while i < n and j < m :
        if arr1[i]<arr2[j]:
            add(arr1[i])
            i+=1
        elif arr1[i]>arr2[j]:
            add(arr2[j])
            j+=1
        else:    #if both elmt r equal , increase i for 1
            i+=1


    while i < n:
        add(arr1[i])
        i+=1
    while j < m:
        add(arr2[j])
        j+=1
    return un
```

```python
ar1=[1,2,3,3,4,7,8,9,9]
ar2=[3,3,4,5,6]
print(union(ar1,ar2))
```
